```python
from sklearn.datasets import load_digits
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### **Step 1: Dataset Loading**

```python
digits = load_digits()
X = pd.DataFrame(digits.data)
y = digits.target

print(f"Dataset shape: {X.shape}")
print(f"Classes: {np.unique(y)}")
print("Data types: ",X.dtypes.tolist())


fig, axes = plt.subplots(1, 5, figsize=(10, 3))
for i, ax in enumerate(axes):
    ax.imshow(digits.images[i], cmap='gray')
    ax.set_title(f"Label: {y[i]}")
    ax.axis('off')
plt.show()
```

```python
X.head()
```

### **Step 2: Data Cleaning**

```python
print("Missing values:", X.isnull().sum().sum())


print("Duplicate rows:", X.duplicated().sum())


print("All features are numeric.")
```

```python
X.corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), cmap='coolwarm', annot=False)
plt.title('Feature Correlation Heatmap')
plt.show()
```

### **Step 3: Filter Method (Feature Selection)**

```python
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0)
X_filtered = selector.fit_transform(X)

print(f"Features before: {X.shape[1]}")
print(f"Features after removing constant pixels: {X_filtered.shape[1]}")
```

### **Step 4: PCA (Unsupervised Feature Extraction)**

### **Code Explanation: Step 4**
- `PCA().fit(X)`: We calculate Principal Component Analysis, which finds the most important 'directions' of variance in the data.
- `np.cumsum(pca.explained_variance_ratio_)`: This calculates how much total information we keep as we add more dimensions.
- `PCA(n_components=10)`: We reduce the 64 original pixels down to just 10 mathematical components.

```python
from sklearn.decomposition import PCA

pca = PCA().fit(X)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA Explained Variance Curve')
plt.grid(True)
plt.show()
```

```python
pca_10 = PCA(n_components=10)
X_pca = pca_10.fit_transform(X)
print(f"Projected data shape via PCA: {X_pca.shape}")
```

### **Step 5: LDA (Supervised Feature Extraction)**

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA

lda = LDA(n_components=9)
X_lda = lda.fit_transform(X, y)

print(f"Projected data shape via LDA: {X_lda.shape}")
```

### **Step 6: Compare Model Performance**

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import time

def evaluate_knn(data, target, description):
    X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.3, random_state=42)
    knn = KNeighborsClassifier(n_neighbors=5)
    start = time.time()
    knn.fit(X_train, y_train)
    acc = accuracy_score(y_test, knn.predict(X_test))
    end = time.time()
    print(f"{description} - Accuracy: {acc:.4f}, Time: {end-start:.4f}s")

evaluate_knn(X, y, "Raw Data (64 features)")
evaluate_knn(X_pca, y, "PCA Data (10 features)")
evaluate_knn(X_lda, y, "LDA Data (9 features)")
```

### **Step 7: Visualize Reduced Feature Space**

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='tab10', alpha=0.7)
plt.title("PCA: First 2 Components")
plt.colorbar(label='Digit Class')

plt.subplot(1, 2, 2)
plt.scatter(X_lda[:, 0], X_lda[:, 1], c=y, cmap='tab10', alpha=0.7)
plt.title("LDA: First 2 Components")
plt.colorbar(label='Digit Class')

plt.tight_layout()
plt.show()
```

### Step 8: Summary
*   **Filter methods** removed pixels that never contain data.
*   **PCA** reduced features while keeping variance, but didn't consider labels.
*   **LDA** performed exceptionally well because it specifically looks for the best way to separate the digit classes.
*   **Conclusion**: LDA is often superior for classification tasks when labels are available, as it achieves higher accuracy with significantly fewer dimensions.
